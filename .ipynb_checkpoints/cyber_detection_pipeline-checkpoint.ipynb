## 3. Audit Data Leakage

> ⚠️ **Obligatoire avant tout entraînement.**  
> Une feature corrélée à >0.95 avec le label = résultats invalides en production.

**Problèmes connus dans ce dataset :**
- `file_path` contient `/Malicious/` ou `/Benign/` dans le chemin → leakage direct 100%
- `file_name` peut encoder la classe (ex: `benign_sample_xxxx`)
- Word : 30 colonnes `path_*/struct_*` avec NaN structurels (10k lignes = format `.doc` legacy)


```python
# Colonnes à supprimer SYSTÉMATIQUEMENT avant tout traitement
LEAKAGE_COLS = ['file_path', 'file_name']

def audit_leakage(X: pd.DataFrame, y: pd.Series, threshold: float = 0.90) -> list:
    """
    Calcule la corrélation point-bisériale de chaque feature avec le label.
    Toute feature avec |r| > threshold est suspecte (leakage potentiel).
    """
    suspicious = []
    for col in X.select_dtypes(include=[np.number]).columns:
        try:
            r, p = pointbiserialr(y, X[col].fillna(0))
            if abs(r) > threshold:
                suspicious.append({"feature": col, "correlation": round(r, 4), "p_value": p})
        except Exception:
            pass
    
    if suspicious:
        print(f"⚠️  {len(suspicious)} feature(s) suspecte(s) (|r| > {threshold}):")
        for s in sorted(suspicious, key=lambda x: -abs(x['correlation'])):
            print(f"   {s['feature']:50s}  r={s['correlation']:+.4f}")
    else:
        print(f"✅ Aucune feature suspecte (seuil |r| > {threshold})")
    
    return suspicious


# Audit rapide sur tous les formats
print("=" * 60)
for fmt, path in CSV_PATHS.items():
    if not path.exists():
        print(f"[{fmt}] fichier manquant — skip")
        continue
    df = pd.read_csv(path)
    X_raw = df.drop(columns=[c for c in LEAKAGE_COLS + ['label'] if c in df.columns])
    y_raw = df['label']
    print(f"\n[{fmt.upper()}] — {len(df)} lignes, {len(X_raw.columns)} features")
    audit_leakage(X_raw, y_raw, threshold=0.90)
print("=" * 60)
```

## 4. Chargement et nettoyage par format

Chaque format est traité indépendamment pour éviter toute contamination croisée.

```python
def load_and_clean(path: Path, fmt: str) -> tuple[pd.DataFrame, pd.Series]:
    """
    Charge un CSV de features, supprime les colonnes de leakage,
    gère les NaN structurels (Word), et retourne (X, y).
    """
    df = pd.read_csv(path)
    print(f"[{fmt}] Chargé : {df.shape} — label dist : {df['label'].value_counts().to_dict()}")
    
    # 1. Supprimer colonnes de leakage
    cols_to_drop = [c for c in LEAKAGE_COLS + ['label'] if c in df.columns]
    X = df.drop(columns=cols_to_drop)
    y = df['label'].copy()
    
    # 2. Gérer NaN structurels (Word : doc legacy vs docx)
    null_count = X.isnull().sum().sum()
    if null_count > 0:
        null_cols = X.columns[X.isnull().any()].tolist()
        print(f"   ⚠ {null_count} NaN dans {len(null_cols)} colonnes")
        
        # Feature binaire : 1 si le fichier est un .doc legacy (NaN dans path_* cols)
        path_cols = [c for c in null_cols if c.startswith('path_') or c.startswith('struct_')]
        if path_cols:
            X['is_legacy_format'] = X[path_cols[0]].isnull().astype(int)
            legacy_dist = X.groupby('is_legacy_format')[[]].apply(
                lambda g: y[g.index].value_counts().to_dict()
            )
            print(f"   is_legacy_format: {X['is_legacy_format'].value_counts().to_dict()}")
        
        # Imputer avec 0 (absence de feature = valeur nulle, sémantiquement correct)
        X = X.fillna(0)
        print(f"   ✅ NaN imputés à 0")
    
    # 3. Vérifier qu'il ne reste que des numériques
    non_num = X.select_dtypes(exclude=[np.number]).columns.tolist()
    if non_num:
        print(f"   ⚠ Colonnes non-numériques ignorées : {non_num}")
        X = X.drop(columns=non_num)
    
    print(f"   ✅ X final : {X.shape}, y : {y.shape}")
    return X, y


# Charger tous les formats
datasets = {}
for fmt, path in CSV_PATHS.items():
    if path.exists():
        print(f"\n{'─'*50}")
        datasets[fmt] = load_and_clean(path, fmt)
    else:
        print(f"[{fmt}] ✗ fichier manquant — ignoré")

print(f"\n✅ Formats chargés : {list(datasets.keys())}")
```

## 5. Extraction de Features QR Code

Les QR codes contiennent des URLs encodées. On extrait deux niveaux de features :

1. **Features image** : entropie visuelle, version QR, modules, résolution
2. **Features URL** (payload décodé) : longueur, entropie, structure du domaine, indicateurs de phishing

> Le CSV associé (`all_generated_urls_*.csv`) contient les URLs générées.  
> On joint les features image + features URL sur le nom de fichier.


```python
# ─── Helpers URL ───────────────────────────────────────────

def shannon_entropy(text: str) -> float:
    if not text:
        return 0.0
    probs = [n / len(text) for n in Counter(text).values()]
    return -sum(p * math.log2(p) for p in probs)

def has_ip_address(url: str) -> int:
    return int(bool(re.search(
        r'(\d{1,3}\.){3}\d{1,3}', urlparse(url).netloc
    )))

def extract_url_features(url: str) -> dict:
    """Extrait ~25 features structurelles d'une URL."""
    if not isinstance(url, str) or not url.strip():
        return {k: 0 for k in URL_FEATURE_NAMES}
    
    url = url.strip()
    try:
        parsed = urlparse(url)
    except Exception:
        return {k: 0 for k in URL_FEATURE_NAMES}
    
    hostname = parsed.hostname or ''
    path     = parsed.path or ''
    query    = parsed.query or ''
    
    # TLD extraction
    if TLDEXTRACT_OK:
        ext = tldextract.extract(url)
        subdomain = ext.subdomain
        domain    = ext.domain
        tld       = ext.suffix
        num_subdomains = len(subdomain.split('.')) if subdomain else 0
    else:
        parts = hostname.split('.')
        tld   = parts[-1] if parts else ''
        domain = parts[-2] if len(parts) >= 2 else hostname
        subdomain = '.'.join(parts[:-2]) if len(parts) > 2 else ''
        num_subdomains = len(parts) - 2 if len(parts) > 2 else 0
    
    query_params = parse_qs(query)
    
    return {
        'url_length':             len(url),
        'url_entropy':            round(shannon_entropy(url), 4),
        'uses_https':             int(parsed.scheme == 'https'),
        'has_ip_address':         has_ip_address(url),
        'has_at_symbol':          int('@' in url),
        'has_double_slash':       int('//' in path),
        'has_port':               int(parsed.port is not None),
        'num_dots':               url.count('.'),
        'num_hyphens':            url.count('-'),
        'num_underscores':        url.count('_'),
        'num_slashes':            url.count('/'),
        'num_question_marks':     url.count('?'),
        'num_equals':             url.count('='),
        'num_ampersands':         url.count('&'),
        'num_percent':            url.count('%'),
        'num_digits_in_url':      sum(c.isdigit() for c in url),
        'num_special_chars':      sum(c in '!#$%^*[]{}|\\<>' for c in url),
        'domain_length':          len(domain),
        'hostname_length':        len(hostname),
        'path_length':            len(path),
        'query_length':           len(query),
        'num_query_params':       len(query_params),
        'num_subdomains':         num_subdomains,
        'tld_length':             len(tld),
        'hostname_digit_ratio':   round(sum(c.isdigit() for c in hostname) / max(len(hostname), 1), 4),
        'has_suspicious_tld':     int(tld.lower() in {
            'tk', 'ml', 'ga', 'cf', 'gq', 'pw', 'top', 'xyz', 'click', 'link', 'work'
        }),
        'has_phishing_keyword':   int(bool(re.search(
            r'(login|signin|verify|account|secure|update|confirm|bank|paypal|amazon|apple)',
            url, re.IGNORECASE
        ))),
        'url_has_hex_encoding':   int(bool(re.search(r'%[0-9a-fA-F]{2}', url))),
        'url_has_shortener':      int(bool(re.search(
            r'(bit\.ly|tinyurl|goo\.gl|t\.co|ow\.ly|is\.gd|buff\.ly)',
            url, re.IGNORECASE
        ))),
    }

URL_FEATURE_NAMES = list(extract_url_features('http://example.com').keys())
print(f"✅ {len(URL_FEATURE_NAMES)} features URL définies :")
print(URL_FEATURE_NAMES)
```

```python
# ─── Helpers IMAGE QR ──────────────────────────────────────

def extract_qr_image_features(img_path: str) -> dict:
    """
    Extrait des features visuelles du PNG QR code.
    Ne nécessite PAS de décoder le QR — analyse l'image brute.
    """
    base = {
        'qr_decode_success':  0,
        'qr_image_width':     0,
        'qr_image_height':    0,
        'qr_image_entropy':   0.0,
        'qr_dark_ratio':      0.0,   # ratio de modules sombres
        'qr_quiet_zone_ok':   0,     # marge blanche détectée
        'qr_version_est':     0,     # version estimée depuis la taille
    }
    
    if not QR_LIBS_OK:
        return base
    
    try:
        img_pil = Image.open(img_path).convert('L')  # grayscale
        w, h = img_pil.size
        base['qr_image_width']  = w
        base['qr_image_height'] = h
        
        # Entropie de l'image (complexité visuelle)
        pixels = list(img_pil.getdata())
        base['qr_image_entropy'] = round(shannon_entropy(str(pixels)), 4)
        
        # Ratio de pixels sombres (modules noirs vs blancs)
        arr = np.array(img_pil)
        base['qr_dark_ratio'] = round(float((arr < 128).mean()), 4)
        
        # Version estimée : taille = (4*version + 17) modules → version = (size-17)/4
        # On estime depuis la taille en pixels (approximatif)
        if w > 0:
            est_modules = w  # si 1px par module
            est_version = max(1, round((est_modules - 17) / 4))
            base['qr_version_est'] = min(est_version, 40)
        
        # Quiet zone : les 4 pixels de bordure doivent être blancs
        border = np.concatenate([arr[0,:], arr[-1,:], arr[:,0], arr[:,-1]])
        base['qr_quiet_zone_ok'] = int(border.mean() > 200)
        
        # Tenter le décodage
        decoded = qr_decode(img_pil)
        if decoded:
            base['qr_decode_success'] = 1
    
    except Exception:
        pass
    
    return base

QR_IMAGE_FEATURE_NAMES = list(extract_qr_image_features.__wrapped__
    if hasattr(extract_qr_image_features, '__wrapped__') else
    {k: 0 for k in ['qr_decode_success','qr_image_width','qr_image_height',
                     'qr_image_entropy','qr_dark_ratio','qr_quiet_zone_ok','qr_version_est']}.keys()
)
print(f"✅ Features image QR : {list(extract_qr_image_features('').keys()) if not QR_LIBS_OK else 'OK (run extract_qr_image_features pour voir)'}")
print(f"✅ Features URL QR   : {len(URL_FEATURE_NAMES)} features")
```

```python
# ─── Pipeline extraction QR complète ────────────────────────

def load_qr_csv(csv_path: Path) -> pd.DataFrame:
    """
    Charge le CSV des URLs générées pour les QR codes.
    Gère plusieurs formats possibles de colonnes.
    """
    df = pd.read_csv(csv_path)
    print(f"QR CSV colonnes : {list(df.columns)}")
    print(df.head(3).to_string())
    
    # Normaliser : on cherche une colonne URL et une colonne filename/index
    url_col = None
    for candidate in ['url', 'URL', 'payload', 'data', 'content', 'link']:
        if candidate in df.columns:
            url_col = candidate
            break
    if url_col is None:
        # Prendre la première colonne non-numérique
        for c in df.columns:
            if df[c].dtype == object and df[c].str.startswith('http', na=False).any():
                url_col = c
                break
    
    if url_col is None:
        raise ValueError(f"Impossible de trouver la colonne URL dans {csv_path}. Colonnes : {list(df.columns)}")
    
    df = df.rename(columns={url_col: 'url'})
    print(f"✅ Colonne URL identifiée : '{url_col}'")
    return df


def build_qr_dataset(
    benign_dir: Path, malicious_dir: Path,
    benign_csv: Path, malicious_csv: Path,
    max_per_class: int = 15000,
    seed: int = SEED
) -> tuple[pd.DataFrame, pd.Series]:
    """
    Construit le dataset QR en joinant :
    - Features URL extraites du CSV
    - Features image extraites des PNG (si disponibles)
    
    Paramètres
    ----------
    max_per_class : limiter le nombre de samples (dataset QR très grand)
    """
    rows = []
    
    for label, url_csv, png_dir in [
        (0, benign_csv, benign_dir),
        (1, malicious_csv, malicious_dir)
    ]:
        class_name = 'benign' if label == 0 else 'malicious'
        
        if not url_csv.exists():
            print(f"⚠ CSV manquant : {url_csv} — génération features URL depuis PNG uniquement")
            url_df = None
        else:
            url_df = load_qr_csv(url_csv)
        
        # Lister les PNG
        png_files = []
        if png_dir.exists():
            png_files = sorted(png_dir.glob('*.png'))[:max_per_class]
            print(f"[{class_name}] {len(png_files):,} PNG trouvés (limité à {max_per_class:,})")
        else:
            print(f"⚠ Dossier PNG manquant : {png_dir}")
        
        if url_df is not None and len(png_files) == 0:
            # Pas d'images — utiliser seulement le CSV
            print(f"[{class_name}] Mode CSV-only ({len(url_df):,} URLs)")
            sample_df = url_df.sample(min(max_per_class, len(url_df)), random_state=seed)
            for _, row in sample_df.iterrows():
                feat = extract_url_features(str(row.get('url', '')))
                # Ajouter features image nulles
                for k in ['qr_decode_success','qr_image_width','qr_image_height',
                           'qr_image_entropy','qr_dark_ratio','qr_quiet_zone_ok','qr_version_est']:
                    feat[k] = 0
                feat['label'] = label
                rows.append(feat)
        
        else:
            # Mode PNG + CSV : joindre par index/filename
            rng = np.random.default_rng(seed)
            indices = rng.choice(len(png_files), size=min(max_per_class, len(png_files)), replace=False)
            
            for i, idx in enumerate(indices):
                if i % 5000 == 0:
                    print(f"   [{class_name}] {i:,}/{len(indices):,}...")
                
                png_path = png_files[idx]
                
                # URL depuis le CSV (par index de ligne ou nom de fichier)
                url = ''
                if url_df is not None:
                    if idx < len(url_df):
                        url = str(url_df.iloc[idx].get('url', ''))
                    # Essayer aussi de matcher par nom de fichier
                    stem = png_path.stem
                    match = url_df[url_df.apply(
                        lambda r: str(r.get('filename', r.get('file', ''))).replace('.png','') == stem, axis=1
                    )]
                    if len(match) > 0:
                        url = str(match.iloc[0].get('url', url))
                
                # Features
                url_feats = extract_url_features(url)
                img_feats = extract_qr_image_features(str(png_path))
                
                feat = {**url_feats, **img_feats, 'label': label}
                rows.append(feat)
    
    df_qr = pd.DataFrame(rows)
    X_qr = df_qr.drop(columns=['label'])
    y_qr = df_qr['label']
    
    print(f"\n✅ Dataset QR construit : {X_qr.shape}")
    print(f"   Distribution : {y_qr.value_counts().to_dict()}")
    return X_qr, y_qr


# ─── DEMO : tester sur quelques URLs sans PNG ───────────────
demo_urls = [
    'https://paypal.com/signin/verify?token=abc123def456',
    'http://192.168.1.1/admin/login.php?redirect=../../../etc/passwd',
    'https://github.com/anthropics/anthropic-sdk-python',
    'https://bit.ly/3xR9pQm',
    'http://amazon-security-alert.tk/update-account?user=victim@email.com',
]
demo_labels = [1, 1, 0, 1, 1]

print("Demo extraction URL features:")
demo_df = pd.DataFrame([extract_url_features(u) for u in demo_urls])
demo_df['url'] = [u[:60]+'...' if len(u)>60 else u for u in demo_urls]
demo_df['label'] = demo_labels
print(demo_df[['url','url_length','has_ip_address','has_phishing_keyword',
               'has_suspicious_tld','url_entropy','url_has_shortener','label']].to_string())
```

## 6. Split Stratifié Train / Val / Test

- **70% train** — entraînement
- **15% val** — early stopping + sélection hyperparamètres  
- **15% test** — évaluation finale (utilisé **une seule fois**)

Split effectué indépendamment par format pour éviter tout mélange de distributions.


```python
def split_format(X: pd.DataFrame, y: pd.Series,
                 test_size: float = 0.15,
                 val_size: float = 0.15,
                 seed: int = SEED) -> tuple:
    """
    Split stratifié 70/15/15.
    Stratifié = maintient le ratio benign/malicious dans chaque split.
    """
    # 1. Extraire le test set
    sss_test = StratifiedShuffleSplit(n_splits=1, test_size=test_size, random_state=seed)
    train_val_idx, test_idx = next(sss_test.split(X, y))
    
    X_tv, X_test = X.iloc[train_val_idx].reset_index(drop=True), X.iloc[test_idx].reset_index(drop=True)
    y_tv, y_test = y.iloc[train_val_idx].reset_index(drop=True), y.iloc[test_idx].reset_index(drop=True)
    
    # 2. Extraire le val set depuis train+val
    val_relative = val_size / (1 - test_size)  # ~0.176 pour obtenir 15% du total
    sss_val = StratifiedShuffleSplit(n_splits=1, test_size=val_relative, random_state=seed)
    train_idx, val_idx = next(sss_val.split(X_tv, y_tv))
    
    X_train = X_tv.iloc[train_idx].reset_index(drop=True)
    X_val   = X_tv.iloc[val_idx].reset_index(drop=True)
    y_train = y_tv.iloc[train_idx].reset_index(drop=True)
    y_val   = y_tv.iloc[val_idx].reset_index(drop=True)
    
    return X_train, y_train, X_val, y_val, X_test, y_test


splits = {}
print(f"{'Format':<8} {'Train':>8} {'Val':>8} {'Test':>8} {'Train label':>15} {'Test label':>15}")
print("─" * 70)

for fmt, (X, y) in datasets.items():
    X_tr, y_tr, X_val, y_val, X_te, y_te = split_format(X, y)
    splits[fmt] = (X_tr, y_tr, X_val, y_val, X_te, y_te)
    tr_dist = y_tr.value_counts(normalize=True).to_dict()
    te_dist = y_te.value_counts(normalize=True).to_dict()
    print(f"{fmt:<8} {len(X_tr):>8,} {len(X_val):>8,} {len(X_te):>8,}  "
          f"B:{tr_dist.get(0,0):.1%}/M:{tr_dist.get(1,0):.1%}  "
          f"B:{te_dist.get(0,0):.1%}/M:{te_dist.get(1,0):.1%}")
```

## 7. Normalisation

> **Règle absolue :** le `StandardScaler` est fitted uniquement sur `X_train`.  
> `X_val` et `X_test` sont uniquement transformés (pas fittés).  
> Fitter sur train+val+test = data leakage de distribution.


```python
scalers = {}
scaled_splits = {}

for fmt, (X_tr, y_tr, X_val, y_val, X_te, y_te) in splits.items():
    scaler = StandardScaler()
    
    X_tr_s  = scaler.fit_transform(X_tr)   # ← fit ICI seulement
    X_val_s = scaler.transform(X_val)       # ← transform uniquement
    X_te_s  = scaler.transform(X_te)        # ← transform uniquement
    
    scalers[fmt] = scaler
    scaled_splits[fmt] = (X_tr_s, y_tr, X_val_s, y_val, X_te_s, y_te, X_tr.columns.tolist())
    
    print(f"[{fmt}] ✅ Scaler fitted sur {len(X_tr):,} samples — {len(X_tr.columns)} features")
    print(f"       mean range : [{scaler.mean_.min():.3f}, {scaler.mean_.max():.3f}]")
    print(f"       std range  : [{scaler.scale_.min():.3f}, {scaler.scale_.max():.3f}]")
```

## 8. Entraînement — LightGBM avec Early Stopping

Un modèle LightGBM par format. `early_stopping_rounds=50` arrête l'entraînement si l'AUC sur le val set ne progresse pas pendant 50 rounds.


```python
LGB_PARAMS = {
    'objective':         'binary',
    'metric':            'auc',
    'n_estimators':      1000,          # early stopping arrêtera avant si besoin
    'learning_rate':     0.05,
    'num_leaves':        31,            # complexité de l'arbre
    'min_child_samples': 20,            # min samples par feuille — anti-overfit
    'subsample':         0.8,           # 80% des lignes par arbre — bagging
    'colsample_bytree':  0.8,           # 80% des features par arbre
    'reg_alpha':         0.1,           # régularisation L1
    'reg_lambda':        0.1,           # régularisation L2
    'class_weight':      'balanced',    # important pour PDF (léger déséquilibre)
    'random_state':      SEED,
    'verbose':           -1,
    'n_jobs':            -1,
}

models = {}
training_history = {}

for fmt, (X_tr, y_tr, X_val, y_val, X_te, y_te, feat_names) in scaled_splits.items():
    print(f"\n{'='*55}")
    print(f"  Entraînement : {fmt.upper()}")
    print(f"  Train : {len(X_tr):,} | Val : {len(X_val):,} | Features : {len(feat_names)}")
    print(f"{'='*55}")
    
    model = lgb.LGBMClassifier(**LGB_PARAMS)
    
    model.fit(
        X_tr, y_tr,
        eval_set=[(X_val, y_val)],
        callbacks=[
            lgb.early_stopping(stopping_rounds=50, verbose=True),
            lgb.log_evaluation(period=100),
        ]
    )
    
    models[fmt] = model
    
    # Métriques rapides sur val
    val_proba = model.predict_proba(X_val)[:, 1]
    val_auc   = roc_auc_score(y_val, val_proba)
    val_preds = (val_proba >= 0.5).astype(int)
    val_f1    = f1_score(y_val, val_preds)
    
    training_history[fmt] = {
        'best_iteration': model.best_iteration_,
        'val_auc': val_auc,
        'val_f1':  val_f1,
    }
    
    print(f"\n  ✅ best_iteration = {model.best_iteration_}")
    print(f"     Val AUC = {val_auc:.4f}")
    print(f"     Val F1  = {val_f1:.4f}")

print("\n✅ Tous les modèles entraînés.")
```

## 9. Évaluation Finale — Test Set

> ⚠️ **Cette cellule ne doit être exécutée qu'une seule fois.**  
> Le test set n'est pas utilisé pendant l'entraînement ni le tuning.


```python
results = {}

fig, axes = plt.subplots(1, len(scaled_splits), figsize=(5 * len(scaled_splits), 4))
if len(scaled_splits) == 1:
    axes = [axes]

for ax, (fmt, (X_tr, y_tr, X_val, y_val, X_te, y_te, feat_names)) in zip(axes, scaled_splits.items()):
    model = models[fmt]
    
    test_proba = model.predict_proba(X_te)[:, 1]
    test_preds = (test_proba >= 0.5).astype(int)
    
    auc  = roc_auc_score(y_te, test_proba)
    f1   = f1_score(y_te, test_preds)
    prec = precision_score(y_te, test_preds)
    rec  = recall_score(y_te, test_preds)
    
    results[fmt] = {'auc': auc, 'f1': f1, 'precision': prec, 'recall': rec,
                    'n_test': len(y_te)}
    
    print(f"\n{'═'*55}")
    print(f"  {fmt.upper()} — Test set ({len(y_te):,} samples)")
    print(f"  AUC       : {auc:.4f}")
    print(f"  F1        : {f1:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}")
    print(classification_report(y_te, test_preds, target_names=['benign','malicious']))
    
    # Confusion matrix
    cm = confusion_matrix(y_te, test_preds)
    sns.heatmap(cm, annot=True, fmt='d', ax=ax, cmap='Blues',
                xticklabels=['benign','malicious'],
                yticklabels=['benign','malicious'])
    ax.set_title(f'{fmt.upper()}\nAUC={auc:.3f}  F1={f1:.3f}')
    ax.set_ylabel('Vrai label')
    ax.set_xlabel('Prédit')

plt.suptitle('Matrices de confusion — Test Set', fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig(OUTPUT_DIR / 'confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()

# Résumé global
print("\n" + "═"*55)
print("RÉSUMÉ GLOBAL")
print("═"*55)
df_results = pd.DataFrame(results).T.round(4)
print(df_results)
```

## 10. Explicabilité — SHAP

SHAP (SHapley Additive exPlanations) permet de comprendre :
- Quelles features ont le plus contribué à la décision
- Dans quelle direction (vers malicious ou benign)


```python
for fmt, (X_tr, y_tr, X_val, y_val, X_te, y_te, feat_names) in scaled_splits.items():
    model = models[fmt]
    
    # Utiliser un sous-ensemble du test pour la vitesse
    n_explain = min(500, len(X_te))
    X_explain = X_te[:n_explain]
    
    print(f"\n[{fmt.upper()}] Calcul SHAP sur {n_explain} samples...")
    
    explainer   = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X_explain)
    
    # Summary plot
    fig, ax = plt.subplots(figsize=(9, 6))
    shap.summary_plot(
        shap_values, X_explain,
        feature_names=feat_names,
        show=False, max_display=15,
        plot_type='bar'
    )
    plt.title(f'Top 15 features SHAP — {fmt.upper()}')
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / f'shap_{fmt}.png', dpi=150, bbox_inches='tight')
    plt.show()
    
    # Top 5 features
    mean_abs_shap = np.abs(shap_values).mean(axis=0)
    top5_idx = np.argsort(mean_abs_shap)[::-1][:5]
    print(f"  Top 5 features :")
    for i, idx in enumerate(top5_idx):
        print(f"    {i+1}. {feat_names[idx]:45s}  SHAP={mean_abs_shap[idx]:.4f}")
```

## 11. Diagnostic Surapprentissage

Comparer AUC train vs AUC val vs AUC test.  
Un écart > 0.03 entre train et val/test signale un surapprentissage.


```python
print(f"{'Format':<8} {'AUC Train':>10} {'AUC Val':>10} {'AUC Test':>10} {'Delta T-V':>12} {'Statut':>10}")
print("─" * 68)

for fmt, (X_tr, y_tr, X_val, y_val, X_te, y_te, feat_names) in scaled_splits.items():
    model = models[fmt]
    
    auc_tr  = roc_auc_score(y_tr, model.predict_proba(X_tr)[:,1])
    auc_val = roc_auc_score(y_val, model.predict_proba(X_val)[:,1])
    auc_te  = results[fmt]['auc']
    delta   = auc_tr - auc_val
    
    status = "✅ OK" if delta < 0.03 else ("⚠ Overfit modéré" if delta < 0.06 else "🔴 Overfit sévère")
    
    print(f"{fmt:<8} {auc_tr:>10.4f} {auc_val:>10.4f} {auc_te:>10.4f} {delta:>12.4f} {status:>10}")
```

## 13. Fonction d'Inférence — Production

Fonction prête à intégrer dans un pipeline SOC.


```python
def predict_file(file_features: dict, fmt: str) -> dict:
    """
    Prédit si un fichier est malicious ou benign.
    
    Paramètres
    ----------
    file_features : dict de features extraites (depuis ton extracteur)
    fmt : 'pdf' | 'html' | 'word' | 'excel' | 'qr'
    
    Retourne
    --------
    dict avec risk_score, label, top_features_shap
    """
    if fmt not in models:
        raise ValueError(f"Format '{fmt}' non supporté. Disponibles : {list(models.keys())}")
    
    model  = models[fmt]
    scaler = scalers[fmt]
    _, _, _, _, _, _, feat_names = scaled_splits[fmt]
    
    # Construire le vecteur de features dans le bon ordre
    X = np.array([[file_features.get(f, 0) for f in feat_names]], dtype=float)
    X_scaled = scaler.transform(X)
    
    # Prédiction
    risk_score = float(model.predict_proba(X_scaled)[0, 1])
    label = 'malicious' if risk_score >= 0.5 else 'benign'
    
    # SHAP pour cette instance
    explainer   = shap.TreeExplainer(model)
    shap_vals   = explainer.shap_values(X_scaled)[0]
    top_idx     = np.argsort(np.abs(shap_vals))[::-1][:5]
    top_features = [
        {'feature': feat_names[i], 'shap_value': round(float(shap_vals[i]), 4)}
        for i in top_idx
    ]
    
    return {
        'format':       fmt,
        'label':        label,
        'risk_score':   round(risk_score, 4),
        'confidence':   round(abs(risk_score - 0.5) * 2, 4),  # 0=incertain, 1=certain
        'top_features': top_features,
        'action':       'BLOCK' if risk_score > 0.8 else ('REVIEW' if risk_score > 0.5 else 'ALLOW'),
    }


# ─── Demo ───────────────────────────────────
if 'pdf' in models:
    # Simuler un fichier PDF suspect
    demo_pdf = {
        'js_count': 3, 'openaction_count': 2, 'launch_count': 1,
        'entropy_of_streams': 7.2, 'name_obfuscations': 5,
        'objstm_count': 2, 'javascript_count': 2,
    }
    result = predict_file(demo_pdf, 'pdf')
    print("Demo prédiction PDF suspect :")
    print(json.dumps(result, indent=2))
```
